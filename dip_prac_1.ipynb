```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image.jpg")
```

**A. Read and display image**

```python
plt.imshow(img)
plt.axis("off")
plt.show()
```

**A. Read and display image and convert BGR to RGB**

```python
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

**B. Resize the given image**

```python
resized = cv2.resize(img,(300,300))
plt.imshow(cv2.cvtColor(resized,cv2.COLOR_BGR2RGB))
plt.title("Resized Image")
plt.axis("off")
plt.show()

cv2.imwrite("resized.jpg",resized)
```

**C. Convert the color image to grayscale**

```python
img_gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(img_gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
cv2.imwrite("gray_image.jpg",img_gray)
```

**D. Convert image to black and white**

```python
_,bw = cv2.threshold(img_gray,127,255,cv2.THRESH_BINARY)
plt.imshow(bw, cmap="gray")
plt.title("Black and White Image")
plt.axis("off")
plt.show()

cv2.imwrite("black_white.jpg", bw)
```

**E. Draw the image profile**

```python
row = img_gray[img_gray.shape[0] // 2, : ]
plt.plot(row)
plt.title("Image Profile")
plt.xlabel("Pixel Position")
plt.ylabel("Intensity")
plt.grid()
plt.show()
```

**F. Separate image into Red, Green and Blue planes**

```python
b,g,r = cv2.split(img)
plt.figure(figsize=(15,5))


plt.subplot(1,3,1)
plt.imshow(r, cmap="gray")
plt.title("Red Plane")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(g, cmap="gray")
plt.title("Green Plane")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(r, cmap="gray")
plt.title("Blue Plane")
plt.axis("off")

plt.show()

cv2.imwrite("blue_plane.jpg",b)
cv2.imwrite("green_plane.jpg",g)
cv2.imwrite("red_plane.jpg",r)
```

**G. Create color image using Red, Green and Blue planes**

```python
reconstructed = cv2.merge([b,g,r])
cv2.imwrite("reconstructed.jpg", reconstructed)

# Convert BGR to RGB for correct display in matplotlib
plt.imshow(cv2.cvtColor(reconstructed, cv2.COLOR_BGR2RGB))
plt.title("Reconstructed Color Image")
plt.axis("off")
plt.show()
```
